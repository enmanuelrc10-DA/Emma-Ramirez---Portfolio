**Conclusions of the Database Analysis**



 - Customers & Employees (Customer, Employee): Stores buyer demographics and links
each customer to their assigned sales support representative.
- Sales & Invoicing (Invoice, InvoiceLine): Records the store's transactional history. Invoice contains general transaction metadata (invoice date, billing address, country, and total amount), while InvoiceLine breaks down the details of each individual item sold (unit price and quantity).
- Product Catalog (Track, Album, Artist, Genre, MediaType): Houses the music available in the store. Each song (Track) is mapped to its corresponding album, artist, musical genre, and media format.
 - Playlists (Playlist, PlaylistTrack): Organizes tracks into custom playlists using a many-to-many relationship.
