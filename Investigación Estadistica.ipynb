## Descipción de la Data

</justify>

El dataset "winequality-red" se obtuvo de la plataforma www.kaggle.com, sitio web muy conocido por ofrecer conjuntos de datos diversos para el análisis y la práctica de técnicas de ciencia de datos y aprendizaje automático. Este dataset específico está diseñado para ser utilizado en estudios de calidad de vinos tintos.

Origen del Dataset
El dataset proviene de un estudio realizado por el Instituto Portugués de Calidad del Vino. Se utilizó para analizar los factores que influyen en la calidad del vino tinto producido en la región del Valle del Duero en Portugal.

Uso del Dataset
El dataset "winequality-red" se utiliza principalmente para:

-Análisis de calidad del vino: Identificar qué características químicas influyen en la calidad del vino.

-Modelado predictivo: Desarrollar modelos de aprendizaje automático que puedan predecir la calidad del vino basado en sus características químicas.

-Estudios académicos: Sirve como base para proyectos de estudiantes y académicos en el campo de la ciencia de datos, aprendizaje automático y enología.

Descripción de las Variables:
El dataset contiene un total de 12 variables, cada una representando diferentes propiedades químicas del vino tinto, además de la calidad del vino. A continuación se describe cada variable y su significado:

Fixed acidity (acidez fija):
Medida de ácidos no volátiles en el vino.
Unidad: g/dm³.

Volatile acidity (acidez volátil):
Medida de ácidos volátiles que pueden evaporarse fácilmente.
Unidad: g/dm³.

Citric acid (ácido cítrico):
Medida de ácido cítrico en el vino, que puede contribuir a la frescura del sabor.
Unidad: g/dm³.

Residual sugar (azúcar residual):
Cantidad de azúcar que queda después de la fermentación.
Unidad: g/dm³.

Chlorides (cloruros):
Medida de la cantidad de sal en el vino.
Unidad: g/dm³.

Free sulfur dioxide (dióxido de azufre libre):
Cantidad de SO2 que no está ligado a otras moléculas y puede actuar como conservante.
Unidad: mg/dm³.

Total sulfur dioxide (dióxido de azufre total):
Cantidad total de SO2, tanto libre como combinado.
Unidad: mg/dm³.

Density (densidad):
Densidad del vino, que puede influir en su cuerpo.
Unidad: g/cm³.

PH:
Medida de acidez/alcalinidad del vino.
Unidad: Escala de pH.

Sulphates (sulfatos):
Indicador de la adición de sales que pueden contribuir a la conservación del vino y mejorar su sabor.
Unidad: g/dm³.

Alcohol:
Contenido de alcohol en el vino.
Unidad: % vol.

Quality (calidad):
Puntuación de calidad del vino dada por expertos en vino.
Escala: 0-10 (variable ordinal).

Este dataset proporciona una base rica para explorar la relación entre las características químicas del vino tinto y su calidad. Se puede utilizar para análisis descriptivos, inferenciales y predictivos, proporcionando una excelente oportunidad para aplicar técnicas de ciencia de datos y aprendizaje automático en el contexto de la enología.
